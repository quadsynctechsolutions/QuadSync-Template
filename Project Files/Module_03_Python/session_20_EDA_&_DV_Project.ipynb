#### E-Commerce EDA & DV Project

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"session_20_ecommerce_eda_dataset.csv")
df.head()
```

```python
# Task 1. Convert a column to a NumPy array

age_array = np.array(df["age"].dropna())
print("Data type:", age_array.dtype)
print("Shape:", age_array.shape)
print("Type",type(age_array))
```

```python
# Task 2. Mean, median and standard deviation

print("Mean age:", np.mean(age_array))
print("Median age:", np.median(age_array))
print("Std deviation:", np.std(age_array))
```

```python
# Task 3. Minimum and maximum value

print("Min price:", np.min(age_array))
print("Max price:", np.max(age_array))
```

```python
# Task 4. Percentiles

print(np.percentile(age_array, 25))
print(np.percentile(age_array, 50))
print(np.percentile(age_array, 75))
```

```python
# Task 5. Detect outliers using IQR

unit_price_array = np.array(df["unit_price"].dropna())
q1 = np.percentile(unit_price_array, 25)
q3 = np.percentile(unit_price_array, 75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outliers = unit_price_array[(unit_price_array < lower) | (unit_price_array > upper)]
print("Number of outliers:", len(outliers))
```

```python
# Task 6. Conditional labeling with np.where()

quantity_array = np.array(df["quantity"].fillna(0))
order_type = np.where(quantity_array > 2, "Bulk", "Normal")
print(order_type[:20])
```

```python
# Task 7. Vectorized revenue calculation

qty = np.array(df["quantity"].fillna(0))
price = np.array(df["unit_price"].fillna(0))
revenue = qty * price
print("Total revenue (all rows):", np.sum(revenue))
```

```python
# Task 8. Rounding values

discount_array = np.array(df["discount"].dropna())
rounded_discount = np.round(discount_array)
print(rounded_discount[:10])
```

```python
discount_array = np.array(df["discount"].dropna())
rounded_discount = np.round(discount_array)
rounded_discount = pd.Series(rounded_discount)
rounded_discount.head(10)
```

```python
# Task 9. Counting missing values with np.isnan()

missing_value_array = np.array(df["unit_price"])
missing_count = np.isnan(missing_value_array).sum()
print("Missing unit_price values:", missing_count)
```

```python
# Task 10. Min-max normalization

price_array = np.array(df["price"])
normalized = (price_array - np.min(price_array)) / (np.max(price_array) - np.min(price_array))
print(normalized[:10])
```
