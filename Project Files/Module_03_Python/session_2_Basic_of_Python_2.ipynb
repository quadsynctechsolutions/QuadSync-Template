### Type Casting

#### int()

```python
# float >> int

val = 12.50
print(val)
print(type(val))

val = int(12.50)
print(val)
print(type(val))
```

```python
# str >> int

val = "108"
print(val)
print(type(val))

val = int("108")
print(val)
print(type(val))
```

```python
# invalid str >> int

val = int("Hello")
print(val)
print(type(val))
```

```python
# bool >> int

val = True
print(val)
print(type(val))

val = int(True)
print(val)
print(type(val))
```

```python
# invalid >> complex >> int

val = 5+6j
print(val)
print(type(val))

val = int(5+6j)
print(val)
print(type(val))
```

#### float()

```python
# int >> float

val = 12
print(val)
print(type(val))

val = float(12)
print(val)
print(type(val))
```

```python
# str >> float

val = "12"
print(val)
print(type(val))

val = float("12")
print(val)
print(type(val))
```

```python
# bool >> float

val = True
print(val)
print(type(val))

val = float(True)
print(val)
print(type(val))
```

```python
# Invalid >> complex >> float

val = 5+6j
print(val)
print(type(val))

val = float(5+6j)
print(val)
print(type(val))
```

#### complex()

```python
# int >> complex

val = 12
print(val)
print(type(val))

val = complex(12)
print(val)
print(type(val))

val = complex(12, 24)
print(val)
print(type(val))
```

```python
# float >> complex

val = 12.5
print(val)
print(type(val))

val = complex(12.5)
print(val)
print(type(val))

val = complex(12.5, 2.4)
print(val)
print(type(val))
```

```python
# bool >> complex

val = True
print(val)
print(type(val))

val = complex(False)
print(val)
print(type(val))

val = complex(True, True)
print(val)
print(type(val))
```

```python
# str >> complex

val = "5"
print(val)
print(type(val))

val = complex("5")
print(val)
print(type(val))

val = complex("5", "6")
print(val)
print(type(val))
```

#### bool()

```python
# int >> bool

val = 5
print(val)
print(type(val))

val = bool(5)
print(val)
print(type(val))
```

```python
# float >> bool

val = 0.0
print(val)
print(type(val))

val = bool(0.0)
print(val)
print(type(val))
```

```python
# str >> bool

val = "Hello"
print(val)
print(type(val))

val = bool("Hello")
print(val)
print(type(val))
```

```python
# complex >> bool

val = 5+6j
print(val)
print(type(val))

val = bool(5+6j)
print(val)
print(type(val))

val = bool(0j)
print(val)
print(type(val))
```
