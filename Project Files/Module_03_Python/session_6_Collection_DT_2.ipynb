# Collection Data Types

## Tuple 

- ordered
- immutable (unchangable)
- round bracket ()

#### Creating tuple

```python
stu = ("Tushar", "Vaibhav", "Avinash")
print(stu)
print(type(stu))
```

```python
# creating empty tuple

names = ()
print(names)
```

```python
names = ("Tushar")
print(type(names))
```

```python
names = ("Tushar", )
print(type(names))
```

#### Accessing tuple

```python
# we can acess element with the help of index

num = (10, 20, 30, 40, 50, 60, 70, 80)
print(num[1])
```

```python
# negative indexing

print(num[-2])
```

```python
# positive slicing

print(num[1:5])
```

```python
# negative slicing

print(num[-6:-1])
```

#### Change tuple value

```python
# as tuple is immutable we cant change any value in tuple

num = (10, 20, 30, 40)
num[1] = 100
```

```python
# you can convert tuple to list and make the change if required and then convert again in tuple

num = (10, 20, 30, 40)

temp = list(num)
temp[1] = 100

num = tuple(temp)
print(type(num))
print(num)
```

#### loop through tuple

```python
# tuple is iterative data type and hence we can iterate its element using loop

num = (10, 20, 30, 40)

for i in num:
    print(i, end=", ")
```

```python
num = (10, 20, 30, 40)

for i in range(1,len(num)+1):
    print(i, end=", ")
```

#### Checking whether the item exists or not (in operator)

```python
colors = ("red", "green", "blue")
print("blue" in colors)
```

```python
print("black" in colors)
```

#### Tuple length

```python
print(len(colors))
```

#### Add & Remove items

```python
# as tuple is immutable hence wqe can't add or delete any element, once tuple is created its unchangable
```

#### Join two or more tuples

```python
t1 = (10, 20)
t2 = ("a", "b")

print(t1 + t2)
```

#### Tuple methods

```python
colors = ("red", "blue", "green", "black", "white", "white")
```

```python
# tuple.count(element)

print(colors.count("red"))
print(colors.count("white"))
```

```python
# tuple.index(element, [start])

colors = ("red", "white", "blue", "green", "black", "white")

print(colors.index("red"))
print(colors.index("white", 2))
```

```python
print(colors.index("Hello"))
```

## Set

- unordered
- unindexed
- curly barcket {}

#### creating set

```python
colors = {"red", "green", "blue", "black", "white"}
print(colors)
print(type(colors))
```

```python
# return only unique element (removes duplicate)

colors = {"red", "green", "blue", "black", "white", "red"}
print(colors)
```

#### Accessing set

```python
# since set is un-indexed hence we can't access element using index

colors = {"red", "green", "blue", "black", "white"}
print(colors[2])
```

```python
# we can acesss element using loop (for loop)

colors = {"red", "green", "blue", "black", "white"}

for i in colors:
    print(i, end=", ")
```

```python
for i in range(len(colors)+1):
    print(colors[i])
```

#### check if item exists

```python
colors = {"red", "green", "blue", "black", "white"}

print("red" in colors)
```

#### changing item in set

```python
# as set is unordered hence we can't make any change in set

colors = {"red", "green", "blue", "black", "white"}
colors[2] = "pink"
```

#### Add element in set

```python
# set.add(element) >> use to add single element in set

colors = {"red", "green", "blue", "black", "white"}
print(colors)
colors.add("pink")
print(colors)
```

```python
# set.update(set2) >> use to add multiple element of set2 in set

colors = {"red", "green", "blue", "black", "white"}
print(colors)
colors2 = {"pink", "yellow", "orange"}
print(colors2)
colors.update(colors2)
print(colors)
```

#### length of set

```python
colors = {"red", "green", "blue", "black", "white"}
print(len(colors))
```

#### Remove item from set

```python
# set.remove(element) >> removes the element from set (generate error if the element is absent)

colors = {"red", "green", "blue", "black", "white"}
colors.remove("red")
print(colors)
```

```python
colors = {"red", "green", "blue", "black", "white"}
colors.remove("pink")
print(colors)
```

```python
# set.discard(element) >> removes the element from set (never through the error if the element is absent)

colors = {"red", "green", "blue", "black", "white"}
colors.discard("red")
print(colors)
```

```python
colors = {"red", "green", "blue", "black", "white"}
colors.discard("pink")
print(colors)
```

```python
# set.pop() >> remove the random element from set

colors = {"red", "green", "blue", "black", "white"}
print(colors)
print(colors.pop())
print(colors)
print(colors.pop())
print(colors)
print(colors.pop())
print(colors)
```

```python
# set.clear() >> removes all the element fro set

colors = {"red", "green", "blue", "black", "white"}
colors.clear()
print(colors)
```

```python
# det set >> to delete entire set 

colors = {"red", "green", "blue", "black", "white"}
print(colors)
del colors
print(colors)
```

#### Union

```python
# set.union(set2) it returns all the element from first set and second set by removing the duplicates
# >> combines both the set and remove duplicates

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors.union(color2))
```

```python
# union operation can be done by | operator as well

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors | color2)
```

#### Intersection

```python
# set.intersection(set2) >> returns the similar element from both the sets

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors.intersection(color2))
```

```python
# intersection can be perform using & operator

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors & color2)
```

#### Difference

```python
# set.difference(set2, [set3, set4.........]) >> returns the element present in frist set but not in second set

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors.difference(color2))
```

```python
# difference can be performed using - operator

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors - color2)
```

#### Symmetric Difference

```python
# set.symmetric_difference(set2, [set3, set4.......]) >> returns all the element from both set except common of both

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors.symmetric_difference(color2))
```

```python
# it can be perform using ^ operator

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors ^ color2)
```

#### Additional Methods

```python
# set.issubset(set2) >> returns True if all the element of calling set exist in specified set

colors = {"red", "green", "blue", "black", "white"}
color2 = {"red", "black"}

print(color2.issubset(colors))
```

```python
# set.issuperset(set2) >> return True if if all the element of specified set exist in calling set

colors = {"red", "green", "blue", "black", "white"}
color2 = {"red", "black"}

print(colors.issuperset(color2)) 
```

```python
# set.isdisjoint(set2) >> return True if not a single element is common from both set

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange"}

print(colors.isdisjoint(color2))
```

```python
# set.intersection_update() >> update calling set according to intersection of calling and specified set (set1 and set2)

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors)

colors.intersection_update(color2)

print(colors)
```

```python
# set.difference_update() >> update calling set according to difference of calling and specified set (returns element from first set which are not present in second set)

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors)

colors.difference_update(color2)

print(colors)
```

```python
# set.symmetric_difference_update() >> update calling set according to symmetric_difference of calling and specified set (returns all element from both set excluding common one)

colors = {"red", "green", "blue", "black", "white"}
color2 = {"pink", "yellow", "orange", "red", "black"}

print(colors)

colors.symmetric_difference_update(color2)

print(colors)
```
