# Function

- block of code which pperform a specific task
- we can pass any data called parameters to functions when it is called
- function returns the data aas a result to its caller

### Types of functions

- 1. User defined functions
- 2. Built in functions

#### User-defined functions

- functions created by programmers based on the requirement of the program are user-defined functions

```python
# defining functions

# def keyword is use to define the function and is mandatory to create functions in python
# return keyword is use to return the values after the operations but it is optional

# def fun_name(parameters):
#     block of Code
#     return some_value

def add(a,b):
    sum = a + b
    return sum

# calling functions >> request for function execution 

sum = add(4,5)

print(sum)
```

```python
"""
 >> parameters are the values that are provided while defining the functions or during calling the functions

 >> when we provide the values during function calling that are called as actual parameters or arguments 
 >> sum = add(4,5) >> this are arguments or actual parameters

 >> when we provide values during function defination are called as formal parameters
 >> def add(a,b) >> this are formal parameters
"""
```

```python
# return statement

"""
  >> return is flow transfer statement
  >> during execution of function if the return is called the execution stops and the control is transfer outside of function from where the function is called
  >> it is option not compulsory
  >> if we are not writing any return statement then the default return value is None
"""
```

### Types of Argument

#### position argument

```python

"""
- focuses mainly on the position of parameters in the way they are passed to the function
- the number of argument and position of arguments must match at all the times
- There is one to one correspondence between actual and formal parameter

"""
```

```python
def add(num1, num2, num3):
    return num1+num2+num3

print(add(1,2,3))
```

```python
print(add(1,2))
```

```python
print(add(1,2,3,4))
```

#### Keyword argument

```python
"""
- in this case, we pass the value of parameter using the parameters name
- no one to one corresspondence between actual and formal parameters
- position of parameters is not important i.e. you can pass the parameter in any order (all parameter must be given)

Note :- python supports both positional and keyword arguments together but make sure you always write the keyword argument after the positional arguments
"""
```

```python
def wish(name, msg):
    return f"Hello! {name}, {msg}"
    
print(wish(name="Amit", msg="Good Morning"))
print(wish(msg="Good Night", name="Avinash"))
```

```python
def wish(name, num, msg):
    return f"Hello! {name}, happy {num} {msg}"
    
print(wish("Ashish", 16, msg="Morning"))
```

```python
def wish(name, num, msg):
    return f"Hello! {name}, happy {num} {msg}"
    
print(wish("Ashish", msg="Morning", 16))
```

#### Default arguments

```python
# default values are assigned to the positional arguments. usend in absence of explicit values during call
```

```python
def wish(name, msg="Good Morning"):
    return f"Hello! {name},  {msg}"
    
print(wish("Ashish"))
```

```python
def wish(name, msg="Good Morning"):
    return f"Hello! {name},  {msg}"
    
print(wish("Ashish", "Good Night"))
```

```python
def wish(name="Ashish", msg):
    return f"Hello! {name},  {msg}"
    
print(wish("Ashish", "Good Night"))
```

#### Variable Length

```python
"""
- variable length arguments means passing any number of arguments to the function
- There are two types of variable length arguments in a dictionary format
    1. *args used when we pass the variable length positional arguments in the tuple format
    2. **kwargs used when we pass variable length keyword arguments in a dictionary forrmat
"""
```

```python
# *args

def add(*args):
    s = 0
    for i in args:
        s += i
    return s

print(add(10, 20, 30, 40, 50))
```

```python
# you can pass variable length arguments along with positional agruments as below

def add(n, *args):
    s = n
    for i in args:
        s += i
    return s

print(add(5, 10, 20, 30, 40, 50)) # n = 5, *args = 10, 20, 30, 40, 50
```

```python
# if the positional argument follows the variable length argument then we need to give the value of positional arguments as keyword argument

def add(*args, n):
    s = n
    for i in args:
        s += i
    return s

print(add(5, 10, 20, 30, 40, n=50)) #*args = 5, 10, 20, 30, 40, n = 50
```

```python
# *kwargs >> when variable length keyword arguments are passed then the keyword and values are treated as key-value pair of dictionary

def fun(**kwargs):
    for i,j in kwargs.items():
        print(f"{i} : {j}")
        
fun(name="Ajit", last_name="More", age=50, gender="Male")
```
