```python
# 2. print(val1 [, val2]...............)

name = "Amit"
age = 25

print(name, age)
print("My name is", name, "My age is", age)
```

```python
# 3. print(val1 [, val2]........., end=" ")

name = "Amit"
age = 25

print(name, age, end=" ")
```

```python
# 6. print(f"string with {variable}")

name = "Amit"
age = 35
salary = 24000

print(f"Name: {name}, Age: {age}, Salary: {salary}")
```

```python
""" 
7. print("format string"% (value1,value2,...))

The format string can contain any of the following format specifiers:

%i - for integer
%d - for integer
%f - for float
%s - for String type

name, age, salary = "Ajit", 49, 4500

print("Name: %s\nAge: %d\nSalary: %f" % (name, age, salary))

Name: Ajit
Age: 49
Salary: 4500.000000
"""

name = "Amit"
age = 35
salary = 24000

print("Name: %s, Age: %d, Salary: %f" % (name, age, salary))
print()
print(" Name: %s\n Age: %d\n Salary: %f" % (name, age, salary))
```
