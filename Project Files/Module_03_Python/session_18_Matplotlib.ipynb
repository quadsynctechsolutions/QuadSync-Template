## Histogram

```python
print("A histogram is used to understand the distribution of a numerical variable. \n")

print("""
| Attribute   | Purpose                 | Example                  |
| ----------- | ----------------------- | ------------------------ |
| x           | Numerical data          | sales                    |
| bins        | Number of intervals     | bins=10                  |
| range       | Data range              | range=(10000,30000)      |
| density     | Probability density     | density=True             |
| cumulative  | Cumulative distribution | cumulative=True          |
| histtype    | Histogram style         | histtype="bar"           |
| orientation | Direction               | orientation="horizontal" |
| rwidth      | Relative bar width      | rwidth=0.8               |
| color       | Histogram color         | color="blue"             |
| alpha       | Transparency            | alpha=0.7                |
| edgecolor   | Bin borders             | edgecolor="black"        |
| linewidth   | Border thickness        | linewidth=1              |
| label       | Legend                  | label="Sales"            |

      """)
```

```python
sales = [
    12000, 13500, 14200, 15000, 15500,
    16000, 16500, 17000, 17500, 18000,
    18500, 19000, 19500, 20000, 21000,
    22000, 23000, 24000, 25000, 26000,
    28000, 30000
]

plt.figure(figsize=(10, 5))

# plt.hist(sales)
plt.hist(sales, bins=5)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

ax.hist(
    sales,
    bins=5,
    alpha=0.8,
    edgecolor="black",
    linewidth=1,
    range=(10000, 30000)
    # cumulative=True
)

ax.set_title("Distribution of Sales")
ax.set_xlabel("Sales Amount (₹)")
ax.set_ylabel("Frequency")

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

plt.show()
```

```python
# A box plot is used to understand the distribution, spread, median, and outliers of a numerical variable.

sales = [
    12000, 13500, 14200, 15000, 15500,
    16000, 16500, 17000, 17500, 18000,
    18500, 19000, 19500, 20000, 21000,
    22000, 23000, 24000, 25000, 26000,
    28000, 30000
]

plt.figure(figsize=(8, 5))

plt.boxplot(sales)

plt.title("Sales Distribution")
plt.grid()
plt.ylabel("Sales")

plt.show()
```

```python
sales = [
    12000, 13500, 14200, 15000, 15500,
    16000, 16500, 17000, 17500, 18000,
    18500, 19000, 19500, 20000, 21000,
    22000, 23000, 24000, 25000, 26000,
    28000, 30000
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.boxplot(sales)

ax.set_title("Sales Distribution")
ax.set_ylabel("Sales")
ax.grid()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.boxplot(sales,
           patch_artist=True,
           vert=True,
           widths=0.4,
           showmeans=True,
           showfliers=True,
           notch=True)

ax.set_title("Sales Distribution")
ax.set_ylabel("Sales Amount (₹)")
ax.grid(axis="y", linestyle="--", alpha=0.8)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))

ax.boxplot(
    sales,
    vert=False
)

ax.set_xlabel("Sales Amount (₹)")
plt.grid()
ax.set_title("Sales Distribution")

plt.show()
```

## Area Chart

```python
print("""
An Area Chart is similar to a line chart, but the area below the line is filled.

It is useful when you want to emphasize magnitude, volume, or cumulative movement over an ordered sequence or over datetime.

| Attribute   | Purpose                     | Example     |
| ----------- | --------------------------- | ----------- |
| x           | X-axis values               | df["Month"] |
| y1          | Upper boundary              | df["Sales"] |
| y2          | Lower boundary              | 0           |
| where       | Select region to fill       | where=...   |
| color       | Area color                  | "skyblue"   |
| alpha       | Transparency                | 0.4         |
| interpolate | Fill boundary intersections | True        |
| step        | Step-style filling          | "pre"       |
""")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.fill_between(
    df["Month"],
    df["Sales"]
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.fill_between(
    df["Month"],
    df["Sales"],
    alpha=0.4
)

plt.plot(
    df["Month"],
    df["Sales"],
    linewidth=2
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(
    linestyle="--",
    alpha=0.4
)

plt.legend()

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 13000, 18000, 22000, 25000]

plt.figure(figsize=(10, 5))

plt.fill_between(
    months,                  # x
    sales,                   # y1
    0,                       # y2
    where=[True, True, True, True, True, True],
    color="skyblue",         # area color
    alpha=0.4,               # transparency
    interpolate=True,        # interpolate boundary
    step=None                # normal area; can use "pre", "post", "mid"
)

plt.plot(
    months,
    sales,
    color="blue",
    linewidth=2,
    linestyle="-",
    marker="o",
    markersize=6,
    label="Sales"
)

plt.title("Monthly Sales Area Chart", fontsize=16, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales (₹)", fontsize=12)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

plt.legend()
plt.tight_layout()
plt.show()
```

## Stack area chart

```python
print("""
A stacked bar chart is used when you want to compare categories while also showing the composition of each category.
      
| Attribute | Purpose                      |
| --------- | ---------------------------- |
| x         | Categories                   |
| height    | Bar values                   |
| bottom    | Starting position of the bar |
| width     | Bar width                    |
| color     | Bar color                    |
| alpha     | Transparency                 |
| edgecolor | Border                       |
| linewidth | Border thickness             |
| label     | Legend                       |
| align     | Bar alignment                |
      """)
```

```python
import numpy as np
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

product_a = np.array([5000, 6000, 5500, 7000, 9000, 10000])
product_b = np.array([4000, 5000, 4500, 6000, 7000, 8000])
product_c = np.array([3000, 4000, 3000, 5000, 6000, 7000])

plt.figure(figsize=(10, 5))

plt.bar(months, product_a, label="Product A")

plt.bar(
    months,
    product_b,
    bottom=product_a,
    label="Product B"
)

plt.bar(
    months,
    product_c,
    bottom=product_a + product_b,
    label="Product C"
)

plt.title("Monthly Sales by Product")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

product_a = np.array([5000, 6000, 5500, 7000, 9000, 10000])
product_b = np.array([4000, 5000, 4500, 6000, 7000, 8000])
product_c = np.array([3000, 4000, 3000, 5000, 6000, 7000])

total = product_a + product_b + product_c

a_percent = product_a / total * 100
b_percent = product_b / total * 100
c_percent = product_c / total * 100

plt.figure(figsize=(10, 5))

plt.bar(
    months,
    a_percent,
    label="Product A"
)

plt.bar(
    months,
    b_percent,
    bottom=a_percent,
    label="Product B"
)

plt.bar(
    months,
    c_percent,
    bottom=a_percent + b_percent,
    label="Product C"
)

plt.title("Monthly Sales Composition")
plt.xlabel("Month")
plt.ylabel("Percentage (%)")

plt.ylim(0, 100)
plt.legend()

plt.show()
```

## Clustered column chart

```python
print("""
A clustered column chart compares multiple numerical values side-by-side for each category.
      """)
```

```python
import numpy as np
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [12000, 15000, 13000, 18000, 22000, 25000]
profit = [2500, 3200, 2800, 4000, 5200, 6000]

x = np.arange(len(months))

width = 0.4

plt.figure(figsize=(10, 5))

plt.bar(
    x - width/2,
    sales,
    width,
    label="Sales"
)

plt.bar(
    x + width/2,
    profit,
    width,
    label="Profit"
)

plt.xticks(x, months)

plt.title("Monthly Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount (₹)")

plt.legend()

plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [12000, 15000, 13000, 18000, 22000, 25000]
profit = [2500, 3200, 2800, 4000, 5200, 6000]

x = np.arange(len(months))

width = 0.35

plt.figure(figsize=(12, 6))

# Sales bars
bars1 = plt.bar(
    x - width/2,              # x position
    sales,                    # height
    width=width,              # bar width
    color="skyblue",          # bar color
    alpha=0.8,                # transparency
    edgecolor="black",        # border color
    linewidth=1.2,            # border thickness
    align="center",           # alignment
    label="Sales",            # legend label
    hatch="//",               # pattern
    zorder=2                  # drawing order
)

# Profit bars
bars2 = plt.bar(
    x + width/2,
    profit,
    width=width,
    color="orange",
    alpha=0.8,
    edgecolor="black",
    linewidth=1.2,
    align="center",
    label="Profit",
    hatch="..",
    zorder=2
)

# X-axis
plt.xticks(
    x,
    months,
    rotation=0,
    fontsize=10
)

# Titles and labels
plt.title(
    "Monthly Sales vs Profit",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel(
    "Month",
    fontsize=12
)

plt.ylabel(
    "Amount (₹)",
    fontsize=12
)

# Grid
plt.grid(
    axis="y",
    linestyle="--",
    linewidth=0.7,
    alpha=0.4,
    zorder=0
)

# Legend
plt.legend(
    title="Metrics",
    loc="upper left",
    fontsize=10,
    title_fontsize=11,
    frameon=True
)

# Data labels
plt.bar_label(
    bars1,
    padding=3,
    fontsize=9,
    fmt="₹%d"
)

plt.bar_label(
    bars2,
    padding=3,
    fontsize=9,
    fmt="₹%d"
)

# Layout
plt.tight_layout()

plt.show()
```

## Multivariate Analysis

```python
print("""
Multivariate analysis is the analysis of three or more variables simultaneously 
to understand relationships, patterns, or interactions among them.
      """)
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

orders = [120, 145, 135, 170, 210, 235]
sales = [12000, 15000, 13000, 18000, 22000, 25000]
profit = [2500, 3200, 2800, 4000, 5200, 6000]

plt.figure(figsize=(10, 6))

plt.scatter(
    orders,
    sales,
    s=[p / 10 for p in profit],
    color="skyblue",
    alpha=0.7,
    marker="o",
    edgecolors="black",
    linewidths=1,
    label="Monthly Data",
    zorder=2
)

for i in range(len(months)):
    plt.annotate(
        months[i],
        (orders[i], sales[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title(
    "Orders vs Sales with Profit",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Orders")
plt.ylabel("Sales (₹)")

plt.grid(
    linestyle="--",
    alpha=0.4,
    zorder=0
)

plt.legend()

plt.tight_layout()

plt.show()
```
