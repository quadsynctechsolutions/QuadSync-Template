#### Adding elements using list methods

```python
# list.append() >> add element to last position of list

num = [10, 20, 30, 40, 50, 60, 70, 80]
num.append(90)
num
```

```python
# list.insert(position, element) >> add element at specified position

num = [10, 20, 30, 40, 50, 70, 80]
num.insert(5, 60)
num
```

```python
# list.extend() >> add list element to another list (it will add element seperatedly not as list)

fruits1 = ["mango", "papaya", "orange"]
fruitsX = ["banana", "apple"]

fruits1.extend(fruitsX)
print(fruits1)
```

```python
fruits1 = ["mango", "papaya", "orange"]
fruitsX = ["banana", "apple"]

fruits1.append(fruitsX)
print(fruits1)
```

```python
fruits1 = ["mango", "papaya", "orange"]
fruitsX = ["banana", "apple"]

fruits1.insert(0, fruitsX)
print(fruits1)
```

#### Sorting and reversing the list

```python
# list.sort() >> to sort in ascending order (default = ascending)

a = [50, 20, 30, 10, 10, 40]
a.sort()
a
```

```python
# list.sort(reverse=True) >> to sort in ascending order (default = ascending)

a = [50, 20, 30, 10, 10, 40]
a.sort(reverse=True)
a
```

```python
# list.reverse() >> reverse the list

a = [50, 20, 30, 10, 10, 40]
a.reverse()
a
```

#### Creating copy of list

```python
# assigning list 1 to list2 will not create a copy it will just create reference

list1 = [10, 20, 30]
list2 = list1

print(id(list1), id(list2))
```

```python
# changes made in one list will reflect in another

list2.append(40)
print(list1, list2)
```

```python
# list.copy() >> used to create clone/copy of the list

list1 = [10, 20, 30]
list2 = list1.copy()

print(id(list1), id(list2))
```

```python
list2.append(40)
print(list1, list2)
```

```python
# list(list) >> used to create clone/copy of the list

list1 = [10, 20, 30]
list2 = list(list1)

print(id(list1), id(list2))
```

```python
list1.append(40)
print(list1, list2)
```

#### List Comprehension

```python
# it is a short and elegant way of creating new list with the help of loop and optional condition in single line
# list = [i for i in iterable if condition]

list = [x for x in range(1,11)]
print(list)
```

```python
square = [x*x for x in range(1,11)]
print(square)
```

```python
even = [x for x in range(1,21) if x%2==0]
print(even)
```

```python
odd = [x for x in range(1,21) if x%2!=0]
print(odd)
```

#### List unpacking

```python
# it is a process of assigning the elements of list to single or multiple variables/statements

list = [10, 20, 30, 40]
a, b, c, d = list
print(a, b, c, d, sep="\n")
```

```python
list = [10, 20, 30, 40]
a, *b, c = list
print(a, b, c, sep="\n")
```
