```python
print("""
To perform file handling in Python, you can first need to open a file using the `open()` function.\n The syntax is as follows:

syntax: open(filename, mode)

Here filename is the name of the file you want to open, and mode is the mode in which you want \nto open the file. The most commonly used modes are:
- 'r' : Read mode - Opens a file for reading (default mode). The file must exist, otherwise it \nwill raise an error.
- 'w' : Write mode - Opens a file for writing. If the file exists, it will be truncated. If the \nfile does not exist, it will be created.
- 'a' : Append mode - Opens a file for appending. If the file does not exist, it will be created.   

Files you can open are as follows:
- Text files: These are files that contain plain text and can be opened using the 'r', 'w',\n or 'a' modes. Examples include .txt, .csv, and .log files.
- Binary files: These are files that contain binary data and can be opened using the 'rb', 'wb',\n or 'ab' modes. Examples include .jpg, .png, and .pdf files.
      """)
```

```python
print("""
Reading methods in Python:
1. read() - Reads the entire file and returns it as a string.   
2. readline() - Reads a single line from the file and returns it as a string.
3. readlines() - Reads all the lines from the file and returns them as a list

writing methods in Python:
1. write() - Writes a string to the file. If the file already exists, it will be overwritten.
2. writelines() - Writes a list of strings to the file.

appending methods in Python:
1. write() - Writes a string to the end of the file. If the file does not exist, it will be created.
      """)
```

### Reading file

```python
# Opening file in read mode

file = open(r"File Handling/data.txt", 'r')
print(file.read())
file.close()
```

```python
# Reading a single line from the file using readline() method

file = open(r"File Handling/data.txt", 'r')
print(file.readline())
file.close()
```

```python
# Reading data from a file using readlines() method

file = open(r"File Handling/data.txt", 'r')
print(file.readlines())
file.close()
```

```python
# Reading a single line from the file using readline() method

file = open(r"File Handling/data2.txt", 'r')
print(file.readline())
file.close()
```

### writing

```python
# Writing data to a file using write() method

file = open(r"File Handling/data.txt", 'w')
file.write("Hello, World!")
file.close()
```

```python
# creating new file using write

file = open(r"File Handling/data2.txt", 'w')
file.write("Hello, World!")
file.close()
```

```python
# writing data to a file using writelines() method

file = open(r"File Handling/data2.txt", 'w')
# print(file.writelines(["Hello \nkaise ho \naccha hu"]))
print(file.writelines(["Hello \n", "kaise ho \n", "accha hu"]))
file.close()
```

### append

```python
# appending data

file = open(r"File Handling/data.txt", 'a')
file.write("\nHello, World!")
file.close()
```

### using with

```python
with open(r"File Handling/data.txt", 'r') as file:
    print(file.read())
```

### Try - Except 

- If programmer want to handle exception gracefully and generate its own user friendly message then use try - except

- Syntax

        try:           
            block of code
        except:
            message

- code that may cause an error is written inside try block and error handling code is written inside except block


- if an exception occurs, python jumps to the matching except block
- if no exception occurs, the except block is skipped

### Try..........Except...........Else

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter Second number: "))
    print(a/b)
except:
    print("Error: Division by zero")
else:
    print("Run Successfullly")
```

### Try..............Except (Ex1)..............Except (Ex2)

```python
list = [10, 20, "Ram", "Sham"]

try:
    a = list[0]
    b = int(list[5])
    print(a/b)
except ZeroDivisionError:
    print("Error: Division by zero")
except TypeError:
    print("Error: Unsupported format")
except ValueError:
    print("Error: Value error")
except IndexError:
    print("Error: index not found")
else:
    print("All run successfully")
```

### try..........except Exception as e................

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter Second number: "))
    print(a/b)
except Exception as e:
    print(e)
```

```python
list = [10, 20, "Ram", "Sham"]

try:
    a = list[0]
    b = int(list[2])
    print(a/b)
except (ZeroDivisionError, TypeError, ValueError):
    print("Error occurs")
else:
    print("All run successfully")
```

### try.............except.............else...............finally

```python
list = [10, 20, "Ram", "Sham"]

try:
    a = list[0]
    b = int(list[2])
    print(a/b)
except (ZeroDivisionError, TypeError, ValueError):
    print("Error occurs")
else:
    print("All run successfully")
finally:
    print("All completed")
```
