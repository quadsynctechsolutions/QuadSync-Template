# Collection Data Types

## Dictionary

- 0rdered
- mutable (changable)
- indexed collection of pair (key - value)

#### Creating dictionary

```python
info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
print(type(info))
print(info)
```

#### Accessing dictionary items

```python
# to access dictionary items use dictionary name followed by key

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
print(info["Name"])
```

```python
# dict.get(key) >> return value of key from dictionary

info.get("Age")
```

#### Adding element or updating elements in dict

```python
info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
print(info)

info["Address"] = "Maharashtra" #adding element in dict
print(info)
```

```python
# updating (adding all the element of dict2 in dict1)

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
dict2 = {"Address": "Maharashtra", "Phone": 7796148325}

print(info)
print(dict2)

print("\n")

info.update(dict2)

print(info)
print(dict2)
```

#### Change value of dictionary items 

```python
# dict[key] = value >> use to change the value of item in dict

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
info["Name"] = "Ekta"
print(info)
```

#### Loop through a dictionary

```python
# print all key in dictionary one by one

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

for i in info:
    print(i)
```

```python
# dict.keys()

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

for i in info.keys():
    print(i)
```

```python
# print all values in dict

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

for i in info:
    print(info[i])
```

```python
# dict.values()

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

for i in info.values():
    print(i)
```

```python
# dict.items() >> print "key : value" together

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

for i, j in info.items():
    print(i, " : ", j)
```

#### Check if the item exists in dict

```python
# in membership operator

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

print("Age" in info)
```

```python
print("age" in info)
```

#### Dictionary Length

```python
info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

print(len(info))
```

#### Remove items from dictionary

```python
# dict.pop(key) >> remove specified key value pair (element)

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
print(info)

info.pop("Age")
print(info)
```

```python
# dict.popitem() >> remove last key value pair (element)

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
print(info)

info.popitem()
print(info)
```

```python
# del dict[key] >> remove specified key value pair (element)

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
print(info)

del info["Age"]
print(info)
```

```python
# del dict >> remove entire variable or dict

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
print(info)

del info
print(info)
```

```python
# dict.clear() >> remove all elements from list

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
print(info)

info.clear()
print(info)
```

#### Creating copy of dictionary

```python
# assigning one dict to another will not create copy of dict it will just create reference

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
dict2 = info

print(id(info), id(dict2))
```

```python
info["Address"] = "Maharashtra"

print(info)
print(dict2)
```

```python
# dict2 = dict.copy() >> create a copy(clone) of dict

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
dict2 = info.copy()

print(id(info), id(dict2))
```

```python
# dict = dict(dict2) >> create a clone/copy of dict

info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}
dict2 = dict(info)

print(id(info), id(dict2))
```

#### dict.setdefault(key[, value = None])

- returns value of key
- if key is absent then adds (key, value) pair
- if value is not given then None is used as default value

```python
info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

print(info.setdefault('Name'))
print(info)
```

```python
info = {'Name' : "Avinash", "Age": 25, "Salary": 50000}

print(info.setdefault('Address', 'Maharashtra'))
print(info)
```
