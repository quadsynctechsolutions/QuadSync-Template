# Control Flow Statement

## Conditional Statement

```python
# if 

x = 10

if x > 5:
    print("Hello Students")
```

```python
# if-else

x = 19

if x >= 18:
    print("18+")
else:
    print("Underage")
```

```python
# if-elif-else

num = -1
 
if num > 0:
    print("Positive Number")
elif num == 0:
    print("Num is zero")
else:
    print("Negative Number")
```

## Iterative Statement

### while

```python
#while

i = 1

while i <= 10:
    print(i, end=" ")
    i += 1
```

```python
# while >> Guess the number 

Num = 8
guess = 0

while Num != guess:
    guess = int(input("Enter the number : "))
    if Num == guess : print(guess, ": That's the correct number")
    else: print("You are wrong")
```

```python
# while-else

i = 1

while i <10:
    print(i, end=" ")
    i += 1
else:
    print("All done")
```

```python
# while-else

i = 1

while i <10:
    if i == 5: break
    print(i, end=" ")
    i += 1
else:
    print("All done")
```

### for

```python
# for loop

for i in range(1,6):
    print(i)
```

```python
list = ["QuadSync", "Tech", "Solutions", "Chh. Sambhajinagar", "Maharashtra"]

for i in list:
    print(i, end=" ")
```

```python
z = "QuadSync Tech Solutions"

for i in z:
    print(i, end=" ")
```

```python
# for - else

for i in range(1,11,2):
    print(i)
else:
    print("All Done")
```

```python
# for - else

for i in range(1,11,2):
    print(i)
    if i == 5: break
else:
    print("All Done")
```

### break, continue, pass

```python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

```python
i = 1

while i <= 10:
    if i == 6:
        break
    print(i)
    i += 1
```

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
i = 0

while i < 10:
    i += 1
    if i == 5:
        continue
    print(i)
```

```python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```

### Exercise on for loop

```python
##sum of numbers

sum = 0

for i in range(1, 101):
    sum += i
    
print(sum)
```

```python
#print numbers from 50 to 100

for i in range(50, 101):
    print(i, end=", ")
```

```python
# print all the odd numbers from 1 to 30

for i in range(1,30):
    if i % 2 != 0:
        print(i)
```

```python
# print all the even numbers from 1 to 30

for i in range(1,30):
    if i % 2 == 0:
        print(i)
```

### Execises on while loop

```python
# print numbers from 100 to 1

num = 100

while num>0:
    print(num, end=", ")
    num -= 1
```

```python
#print multiples of 7 upto 70

mul = 7

while mul <= 70:
    print(mul)
    mul += 7
```

```python
## find the sum of all even number from 2 to 20

i = 2
sum = 0

while i <=20:
    if i % 2 ==0:
        sum += i
    i += 1
        
print(sum)
```
