# Data Filtering

### DataFrame

```python
data = {
    'EmpID': [101, 102, 103, 104, 105],
    'Name': ['Ajay', 'Neha', 'Ravi', 'Priya', 'Amit'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT'],
    'Salary': [50000, 45000, 70000, 60000, 55000],
    'City': ['Pune', 'Mumbai', 'Pune', 'Delhi', 'Mumbai']
}

df = pd.DataFrame(data)

print(df)
```

##### Filter row based on only one condition

```python
print(df['Salary'] > 55000) # only return boolean series
```

```python
print(df[(df['Salary'] > 55000)])
```

##### Filter row based on multiple condition

- we have to use bitwise (boolean) as logical not works in Pandas

```python
# AND (&) << Boolean (bitwise AND)

print(df[
        (
            (df['Department'] == 'IT') 
        & 
            (df['Salary'] > 50000)
        )
    ])
```

```python
# OR (|) << Boolean (bitwise OR)

print(df[((df['Department'] == 'IT') | (df['Salary'] > 50000))])
```

```python
# NOT (~) << Boolean (bitwise NOT)

print(df[~(df['Department'] == 'IT')])
```

##### Filtering using isin()

```python
# Employee staying in pune or mumbai

print(df[df['City'].isin(['Pune', 'Mumbai'])])
```

##### Filter using between()

```python
print(df[df['Salary'].between(50000, 60000)])
```

##### Filter using string conditions

```python
# Starts with

print(df[df['Name'].str.startswith('A')])
```

```python
# Ends with

print(df[df['Name'].str.endswith('a')])
```

##### Combining DataFrame

        pd.concat(object, axis = 0, ignore_index = False, Keys = None)

```python
import pandas as pd

jan = pd.DataFrame({'Month':   ['Jan', 'Jan', 'Jan'],
    		    	'Product': ['Laptop', 'Phone', 'Tablet'],
    		    	'Sales':   [120000, 45000, 30000]})

feb = pd.DataFrame({'Month':   ['Feb', 'Feb', 'Feb'],
    	            'Product': ['Laptop', 'Phone', 'Tablet'],
                    'Sales':   [135000, 52000, 28000]})

mar = pd.DataFrame({'Month':   ['Mar', 'Mar', 'Mar'],
    			    'Product': ['Laptop', 'Phone', 'Tablet'],
    			    'Sales':   [110000, 61000, 35000]})

print(jan)
print()
print(feb)
print()
print(mar)
print()
```

```python
# Stack all three vertically
combined = pd.concat([jan, feb, mar], ignore_index=True)
print(combined)
```

```python
combined = pd.concat([jan, feb, mar], ignore_index=False)
combined2 = pd.concat([jan, feb, mar]) # default value is false
print(combined)
print()
print(combined2)
```

```python
# vertical stacking (axis=0) << By default Value

combined = pd.concat([jan, feb, mar], axis = 0, ignore_index=True)
print(combined)
```

# Time Series

##### Basic

```python
print("""
A time series is a collection of data recorded in time order, usually at regular intervals such as every hour, day, month, or year.

A time series is a sequence of data points or observations recorded in chronological order over time, usually at regular time intervals.

A Time Series is a sequence of observations recorded over time.

Examples:
● Daily stock prices
● Monthly sales
● Weather data
● Website visitors
● Electricity consumption

In Pandas, dates are stored using DatetimeIndex, which makes time-based analysis very easy.
      """)
```

```python
data = {
 "Date": [
 "2025-01-01",
 "2025-01-02",
 "2025-01-03",
 "2025-01-04",
 "2025-01-05",
 "2025-01-06",
 "2025-01-07"
 ],
 "Sales": [120, 135, 128, 150, 145, 160, 170]
}

df = pd.DataFrame(data)

df
```

```python
df["Date"] = pd.to_datetime(df["Date"])
print(df.dtypes)
```

##### Extract Date Components

```python
print(df.index.year)
print()
print(df.index.month)
print()
print(df.index.day)
print() 
print(df.index.day_name())
print()
print(df.index.month_name())
```

##### Rolling Window


```python
# Calculates moving statistics.
# Rolling Sum
# create windows of size 3 and calculate the sum of each window

df["Sales"].rolling(3).sum()
```

##### Expanding Window

```python
# Calculates cumulative statistics.
# Cumulative Sum
# Cumulative Sum calculates the sum of all previous values in a column, including the current value.

df["Sales"].expanding().sum()
```

##### Shifting Data

```python
# Previous Day Sales
# The shift() function is used to shift the values of a column up or down by a specified number of periods. In this case, we are shifting the "Sales" column down by 1 period to get the previous day's sales.

df["Previous"] = df["Sales"].shift(1)

df
```

```python
# Next Day Sales
# The shift() function is used to shift the values of a column up or down by a specified number of periods. In this case, we are shifting the "Sales" column up by -1 period to get the next day's sales.

df["Next"] = df["Sales"].shift(-1)
df
```

##### Percentage Change

```python
# Shows growth rate.
# The pct_change() function is used to calculate the percentage change between the current and previous elements in a column. In this case, we are calculating the percentage change in sales from one day to the next.

df['pct_change'] = df["Sales"].pct_change()*100
df
```

##### Difference Between Consecutive Values

```python
# The diff() function is used to calculate the difference between the current and previous elements in a column. In this case, we are calculating the difference in sales from one day to the next.

df["Sales"].diff()
```

##### Date Range

```python
# Create a date range.
# The date_range() function is used to create a range of dates with a specified start date, number of periods, and frequency. In this case, we are creating a date range starting from "2025-01-01" for 7 periods with a daily frequency.

date = pd.date_range(
 start="2025-01-01",
 periods=7,
 freq="D"
)

date = date.to_list()

for i in date:
    print(i.date())
```

##### Handling Missing Dates

```python
import pandas as pd
```

```python
data = {
 "Date": ["2025-01-01", "2025-01-04", "2025-01-08", "2025-01-12", "2025-01-16", "2025-01-20", "2025-01-24"],
 "Sales": [120, 135, 128, 150, 145, 160, 170]
}

df = pd.DataFrame(data)

df["Date"] = pd.to_datetime(df["Date"])

df.set_index("Date", inplace=True)
print(df)
print(df.dtypes)
```

```python
df = df.asfreq("D")
df
```

```python
# Fill missing values
# The fillna() function is used to fill missing values in a DataFrame. In this case, we are using the method parameter to specify that we want to forward fill the missing values, which means that we will fill the missing values with the last known value.

df.fillna(method="ffill")
```

```python
df.ffill() 
```

```python
# Backward fill
# The bfill() function is used to fill missing values in a DataFrame. In this case, we are using the method parameter to specify that we want to backward fill the missing values, which means that we will fill the missing values with the next known value.

df = df.bfill()
df
```
