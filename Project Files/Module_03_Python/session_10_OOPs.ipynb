# Procedure Oriented Programming

```python
# Divides problem into sub-task called functions or procedure
# Data to be shared by related functions is kept global i.e. less importance of data
# Less real world mapping

# Top-Down approch
```

# Object - Oriented Programming

```python
# Represents problem in terms of classess and objects
# Encapsulation of related functions and data into classes
# Good real world mapping

# Each function is called as capsule and class consists of multiple capsules called as encapsulation

# Variable derived in class is called as local variable and it can't be use/access outside class

# bottom - up approach
```

```python
# Everything in python is a object

num = 10
print(type(num))
print()

s = "Hello "
print(type(s))
print()

def f():
    print("Hello")

print(type(f))
```

### Adding Methods

```python
''' 
class ClassName:
    def methodName(self [, arr1, arg2.........]):
        method_body
'''

class Cat:
    def set_data(self, height, color):
        self.height = height
        self.color = color
        
    def show(self):
        print(f"Hello, My color is {self.color}, and height is {self.height}")
        
kitty = Cat()
kitty.set_data(2.2, "White") # Cat.set_data(kitty, 2.2, "White") >> Explicitly
kitty.show() # Cat.show(kitty)

mani = Cat()
mani.set_data(2.1, "Black")
mani.show()
```

```python
x = Cat()
x.show()
```

### Constructor

- Special method of class which is called implicitly, immediately after object is instantiated
- use to give initial value to data members of class
- Identified by special magic method __init__

Syntax:

        def __init__(self [, parameters]):
            self......... = ........

- self is compulsory for constructors also because self represents the specific object (instance of class) being created and referred

```python
class Cat:
    
    def __init__(self, height, color = "Black"):
        self.height = height
        self.color = color
    
    def show(self):
        print(f"My height is {self.height}")
        print(f"My color is {self.color}")
        
kitty = Cat(2.2, "White") # implicitly Cat.__init__(kitty, 2.2, "White") is called
kitty.show()

print()

mani = Cat(2.4)
mani.show()
```
